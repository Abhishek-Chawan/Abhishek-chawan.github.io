```python
import random

p = input("ASK:").lower()
k_b = {
    "college": {
        "keywords": ["bengaluru", "hub", "college", "nhce"],
        "response": ["bengaluru is hub of it college and among them nhce is best engineering college"]
    },
    "politics": {
        "keywords": ["modi", "pm", "india"],
        "response": ["modi is current pm of india and he leader of world largest democratic nation"]
    },
    "cricket": {
        "keywords": ["sport", "most", "popular", "virat"],
        "response": ["cricket one of most popular sport and in current virat  best of this sport "]
    }
}

best_intent = None
highest_score = 0
for intent_name, intent_data in k_b.items():

  score  = sum(1 for word in intent_data["keywords"] if word in p)
  if score > highest_score:
    best_intent = intent_name
    highest_score = score

print("\n AI response:")
if best_intent and highest_score >= 2:
  print(random.choice(k_b[best_intent]["response"]))
else:
  print("sorry i dont know")
```
